# Pharmacovigilance pipeline — debug notebook

Inspects **50 random** records: original text, drugs, symptoms, links, onset, reports.

```python
import random
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "debugging":
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from data_loader.loader import load_all
from debugging.display import format_debug
from pipeline import process_record
from symptom_extraction.symptoms import extract_symptoms_batch
```

```python
posts_df, comments_df, post_recs, comment_recs = load_all()
records = post_recs + comment_recs
print(f"posts={len(post_recs)} comments={len(comment_recs)} total={len(records)}")
```

```python
N_SAMPLES = 50
random.seed(42)
sample = random.sample(records, min(N_SAMPLES, len(records)))
symptom_lists = extract_symptoms_batch([r.text for r in sample], batch_size=16)
debugs = [process_record(r, syms)[1] for r, syms in zip(sample, symptom_lists)]
```

```python
for i, dbg in enumerate(debugs, 1):
    print(f"\n### SAMPLE {i}/{len(debugs)} ###")
    print(format_debug(dbg))
```
